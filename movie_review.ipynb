```python
from datasets import load_dataset
import pandas as pd

data = load_dataset("stanfordnlp/imdb")
train = pd.DataFrame(data["train"])
test = pd.DataFrame(data["test"])
print(train.shape, test.shape)
train.head()
```

```python
train["text"] = train["text"].str.replace("<br />", " ", regex=False)
test["text"] = test["text"].str.replace("<br />", " ", regex=False)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer(stop_words="english", max_features=50000)
X_train = vectorizer.fit_transform(train["text"])
X_test = vectorizer.transform(test["text"])
y_train = train["label"]
y_test = test["label"]
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

model1 = LogisticRegression(max_iter=1000)
model1.fit(X_train, y_train)
pred1 = model1.predict(X_test)
print("Logistic Regression accuracy:", accuracy_score(y_test, pred1))
```

```python
from sklearn.svm import LinearSVC

model2 = LinearSVC()
model2.fit(X_train, y_train)
pred2 = model2.predict(X_test)
print("Linear SVM accuracy:", accuracy_score(y_test, pred2))
```

```python
wrong = test[pred2 != y_test].head(10)
for _, row in wrong.iterrows():
    print("Actual:", "positive" if row["label"] == 1 else "negative")
    print(row["text"][:300], "\n---")
```
